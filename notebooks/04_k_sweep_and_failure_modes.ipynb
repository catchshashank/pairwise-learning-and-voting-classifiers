```
# Plot of accuracy vs k + a “why it fails” markdown cell.

import matplotlib.pyplot as plt

ks = [1,2,3,4,5,6,7,8,10]
accs = [eval_voting(n=200, k=k) for k in ks]

plt.figure()
plt.plot(ks, accs, marker="o")
plt.xlabel("k (references per class)")
plt.ylabel("Voting Accuracy (sampled)")
plt.title("k-shot Voting Accuracy vs k")
plt.show()
```

```
## Failure modes to look for
- Pairwise model near chance (≈50%) → voting becomes noisy
- Some digits (e.g., 4/9, 3/5) may confuse pairwise similarity more
- Increasing k can help by averaging noise, but can also saturate if the pairwise classifier is weak
```
